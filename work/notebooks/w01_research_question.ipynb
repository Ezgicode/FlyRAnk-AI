## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Research question

Which content pages should be prioritized for human refresh review, based on observable
search and content signals?

### Decision and action

This project aims to improve the decision of which pages an editor should review first.
The output will be a ranked list of pages with a priority score and a short reason code.
An editor can use this list to inspect the highest-priority pages and decide whether to
refresh, expand, improve the click-through context, monitor, merge, or leave a page unchanged.

### Cost of a wrong recommendation

A false positive means that an editor spends time reviewing a page that does not need
 action or does not have enough potential value. A false negative means that an important
  declining or high-opportunity page is missed and may continue to lose visibility or traffic.
 For this reason, the output should support human review rather than automatically changing content.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains **30,000** page-level content records. Of these pages, **16,262 (54.2%)** have a declining trend label in the starter exercise. In addition, **16,726 pages (55.8%)** received at least 500 impressions in the previous 90 days.

These numbers suggest that there is a large enough set of visible pages to prioritize, while not every page should receive the same level of human attention. A ranking approach can help focus review effort on pages with both measurable visibility and signals of possible decline or opportunity.

```python
import os
import sys
import subprocess
from pathlib import Path
import pandas as pd

repo_name = "flyrank-ml-internship-starter"
data_file = Path("data/raw/content_refresh_anonymized.csv")

if not data_file.exists():
    if not Path(repo_name).exists():
        subprocess.run(
            ["git", "clone", "--depth", "1",
             "https://github.com/flyrank-bih/flyrank-ml-internship-starter.git",
             repo_name],
            check=True
        )
    os.chdir(repo_name)
    data_file = Path("data/raw/content_refresh_anonymized.csv")

df = pd.read_csv(data_file)

total_pages = len(df)
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
declining_rate = declining_pages / total_pages
visible_pages = (df["impressions_90d"] >= 500).sum()
visible_rate = visible_pages / total_pages

print(f"Total page-level records: {total_pages:,}")
print(f"Declining pages: {declining_pages:,} ({declining_rate:.1%})")
print(f"Pages with at least 500 impressions in 90 days: {visible_pages:,} ({visible_rate:.1%})")
```
